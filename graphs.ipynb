```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
psnr_fista = np.load('metricas/Fista_psnr.npy')
psnr_ista = np.load('metricas/Ista_psnr.npy')

plt.figure(figsize=(10, 6)) 

# 3. Graficar el PSNR
plt.plot(psnr_fista, color='b', label='FISTA') 
plt.plot(psnr_ista, color='r', label='ISTA')

plt.title('Algorithms comparison: ratio 0.7 + Denoiser', fontsize=16)
plt.ylabel(r'PSNR (dB)', fontsize=14)
plt.xlabel(r'Iteration', fontsize=14)
plt.grid(True) 
plt.legend(fontsize=14)
plt.show()
```

```python
error_fista = np.load('metricas/Fista_error.npy')
error_ista = np.load('metricas/Ista_error.npy')

plt.figure(figsize=(10, 6))
# 3. Graficar el error
plt.plot(error_fista, color='b', label='FISTA')
plt.plot(error_ista, color='r', label='ISTA')

plt.title('Algorithms comparison: ratio 0.7 + Denoiser', fontsize=16)
plt.yscale('log')
plt.ylabel(r'$\frac{1}{2} \|\mathbf{y} - \mathbf{H(x)}\|^2$', fontsize=14)
plt.xlabel(r'Iteration', fontsize=14)
plt.grid(True)
plt.legend(fontsize=14)
plt.show()
```

```python
ratios = [0.3, 0.7]
colors = ['r', 'b']
plt.figure()
for idx, ratio in enumerate(ratios):
    convergence_fista = np.load(f'metricas/Fista_convergence_rate{ratio}.npy')
    psnr_final = np.load(f'metricas/Fista_psnr{ratio}.npy')
    psnr = psnr_final[:,0][-1]
    psnr = np.around(psnr, 4)
    print(psnr)
    plt.plot(convergence_fista[:,0], color=colors[idx], label=f'ratio {ratio}, psnr={psnr}')
#plt.yscale('log')
plt.ylabel(r'$\frac{\|\mathbf{x}^{l} - \mathbf{x}^{*}\|}{\|\mathbf{x}^{l-1} - \mathbf{x}^{*}\|} $', fontsize=14)
plt.xlabel(r'Iteration', fontsize=14)
plt.legend(fontsize=14)
plt.title('FISTA: Convergence rate', fontsize=16)
plt.show()
```
